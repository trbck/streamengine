```python
import numpy as np
import nest_asyncio
import asyncio
import logging
from redisobjstore import RedisObjectStorage

class Stock:
    def __init__(self, ticker, company_name, quotes):
        self.ticker = ticker
        self.company_name = company_name
        self.quotes = quotes

    def __repr__(self):
        return f"Stock(ticker={self.ticker}, company_name={self.company_name}, quotes=Array of shape {self.quotes.shape})"

async def main():
    # Initialize RedisObjectStorage with INFO log level
    storage = RedisObjectStorage(log_level=logging.INFO)

    # Example Stock object
    quotes = np.random.rand(1000, 10)
    stock = Stock(ticker="AAPL", company_name="Apple Inc.", quotes=quotes)

    # Store the Stock object
    await storage.store_with_pickle('stock:AAPL', stock)

    # Retrieve the Stock object
    retrieved_stock = await storage.retrieve_with_pickle('stock:AAPL')
    print('Retrieved with pickle:', retrieved_stock)
    print('Quotes:', retrieved_stock.quotes if retrieved_stock else "No data")

    # List all keys matching a pattern
    keys = await storage.list_keys('stock:*')
    print('Keys:', keys)

    # Delete all keys matching a pattern
    await storage.delete_keys('stock:*')

    # Verify deletion
    keys_after_deletion = await storage.list_keys('stock:*')
    print('Keys after deletion:', keys_after_deletion)

    # Close the Redis client connection
    await storage.close()

# Allow nested use of asyncio.run() in Jupyter Notebook
nest_asyncio.apply()

# Run the main function
asyncio.run(main())
```
